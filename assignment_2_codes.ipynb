```python
import cv2
import matplotlib.pyplot as plt
```

#### Question 1 : Write a Python program to read an image named "flower.jpg" using OpenCV and display it in a window titled "Original Image".

```python
# Loading the image using openCV (loaded in BGR format)
img = cv2.imread('Images/flower.jpg')

# Converting image from BGR to RGB and displaying using Matplotlib
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
plt.show()
```

#### Question 3 : Write a Python program to load a colour image and convert it into a grayscale image, then save the result as “output_gray.jpg”.

#### Question 5 : Write a Python program using OpenCV that resizes an image to 50% of its original width and height.

```python
# Load the image using openCV.
img4 = cv2.imread('Images/flower.jpg')

# Resizing the image to 50% of its original width and height.
h, w = img4.shape[:2]
resized = cv2.resize(img4, (w//2, h//2))

# Displaying the resized image using Matplotlib.
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title('Resized Image')
plt.axis('off')
plt.show()
```

#### Question 6 : Write a Python code snippet to crop the central 200×200 pixel area of an image and display it using Matplotlib.

```python
# Load the image using openCV.
img5 = cv2.imread('Images/flower.jpg')

# Croping the central 200x200 pixel area.
h, w = img5.shape[:2]
crop = img5[h//2-100:h//2+100, w//2-100:w//2+100]

# Displaying the cropped image.
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title('Cropped Image')
plt.axis('off')
plt.show()
```

#### Question 7 : Write a short program that rotates an image by 90 degrees clockwise using OpenCV.

```python
# Load the image using openCV.
img6 = cv2.imread('Images/flower.jpg')

# Rotating the image 90 degrees clockwise.
rotated = cv2.rotate(img6, cv2.ROTATE_90_CLOCKWISE)

# Displaying the rotated image using Matplotlib.
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated Image')
plt.axis('off')
plt.show()
```

#### Question 8 : Write a Python statement to flip the image “car.jpg” left-to-right using OpenCV and display the result.

```python
# Load the image using openCV.
img7 = cv2.imread('Images/car.jpg')

# Fliping the image horizontally.
flipped = cv2.flip(img7, 1)

# Displaying the flipped image using Matplotlib.
plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
plt.title('Horizontal Flip')
plt.axis('off')
plt.show()
```

#### Question 9 : Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
# Load the image using openCV.
img8 = cv2.imread('Images/flower.jpg')

# Converting the image from BGR to HSV.
hsv = cv2.cvtColor(img8, cv2.COLOR_BGR2HSV)

# Spliting the image into Hue, Saturation and Value channels.
h, s, v = cv2.split(hsv)

# Displaying the three channels separately.
plt.figure(figsize=(12,3))
for i, (img, title) in enumerate(zip([h, s, v], ['Hue', 'Saturation', 'Value'])):
    plt.subplot(1,3,i+1)
    plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.show()
```

#### Question 10 : Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, "img1.jpg" and "img2.jpg".

```python
# Import the MSE function.
from skimage.metrics import mean_squared_error

# Load both images in grayscale.
img9 = cv2.imread('Images/img1.jpg', cv2.IMREAD_GRAYSCALE)
img10 = cv2.imread('Images/img2.jpg', cv2.IMREAD_GRAYSCALE)

# Calculate the Mean Squared Error.
mse = mean_squared_error(img9, img10)

# Print the MSE value.
print("Mean Squared Error:", mse)
```
